# संपीड़न (Compression)
प्रदाता सहित भाष कर्नेल चुनें और कक्ष क्रम से चलाएँ। JVM के लिए आपके Linux x64, Windows x64 या Apple Silicon का provider चाहिए। यह उदाहरण इंटरनेट का प्रयोग नहीं करता। बाइटबफ़र फलन के भीतर स्थानीय हैं; पाठ मान अगले कक्ष में भी उपलब्ध है।

```bhash
// bhash:null-safety=strict
आयात मानक.संपीड़न;
आयात मानक.संपीड़नप्रारूप;
```

## गज़िप चक्र (Gzip roundtrip)
पहले UTF-8 बाइटें बनती हैं। संपीड़न की सफलता पर ही विसंपीड़न होता है। दोनों Result की त्रुटियाँ संभाली जाती हैं। १०२४ सीमा output बाइटों की है, अक्षरों की नहीं।

```bhash
फलन चक्र(मूल: पाठ) -> पाठ {
    लौटाओ मिलान (संपीड़न.संपीड़ितकरो(यूटीएफ८बफ़र(मूल), संपीड़नप्रारूप.गज़िप, ६)) {
        विकल्प मानक.परिणाम.असफल(स्थिर दोष) => "त्रुटि: " + दोष.संदेश,
        विकल्प मानक.परिणाम.सफल(स्थिर पैक) => मिलान (संपीड़न.विसंपीड़ितकरो(पैक, संपीड़नप्रारूप.गज़िप, १०२४)) {
            विकल्प मानक.परिणाम.असफल(स्थिर दोष) => "त्रुटि: " + दोष.संदेश,
            विकल्प मानक.परिणाम.सफल(स्थिर खुला) => यूटीएफ८पाठ(खुला),
        },
    };
}
```

```bhash
स्थिर संदेश: पाठ = "नमस्ते भारत 😀";
```

```bhash
चक्र(संदेश)
```

अपेक्षित मान: `नमस्ते भारत 😀`। फिर output सीमा १०२४ से १ करके फलन वाला कक्ष दोबारा चलाएँ। अंतिम कक्ष अब हिन्दी त्रुटि संदेश दिखाएगा।
