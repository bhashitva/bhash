# पहली भाष नोटबुक
Bhash (भाष) कर्नेल चुनें। हर कक्ष को क्रम से चलाएँ।

```bhash
मान कुल: पूर्णांक = ०;
```

```bhash
कुल = कुल + २;
कुल
```

ऊपर वाला कक्ष दोबारा चलाएँ। `कुल` में फिर से २ जुड़ेंगे। कर्नेल पुनः आरम्भ करने पर सत्र के मान मिट जाते हैं।

```bhash
फलन दुगुना(अ: पूर्णांक) -> पूर्णांक { लौटाओ अ * २; }
```

```bhash
प्रतीक्षा मानक.कार्य.विलंब(२५);
दुगुना(कुल)
```
